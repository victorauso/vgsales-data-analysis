### 1. Importação das Bibliotecas e Carga dos Dados

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração estática de visualizações
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

# Carga dos dados
url = "https://raw.githubusercontent.com/gregrahn/join-order-benchmark/master/vgsales.csv"
df = pd.read_csv(url)
print(f"Dataset carregado com {df.shape[0]} linhas e {df.shape[1]} colunas.")
df.head()
```

### 2. Tratamento e Limpeza dos Dados (Data Cleaning)

```python
# Verificando dados nulos
print("--- Valores nulos por coluna ---")
print(df.isnull().sum())

# Removendo registros sem ano de lançamento ou publisher definida
df_clean = df.dropna(subset=['Year', 'Publisher']).copy()
df_clean['Year'] = df_clean['Year'].astype(int)

print(f"\nLinhas após limpeza: {len(df_clean)}")
```

### 3. As 10 Maiores Plataformas de Todos os Tempos (Vendas Globais)

```python
top_plataformas = df_clean.groupby('Platform')['Global_Sales'].sum().sort_values(ascending=False).head(10).reset_index()

sns.barplot(data=top_plataformas, x='Global_Sales', y='Platform', palette='Blues_r')
plt.title('Top 10 Consoles Mais Vendidos na História (em milhões de cópias)', fontsize=14, fontweight='bold')
plt.xlabel('Vendas Globais (Mi de unidades)')
plt.ylabel('Console / Plataforma')
plt.show()
```

**💡 Insight 1:** O **PlayStation 2 (PS2)** lidera como a plataforma com maior volume total de vendas de jogos na história, seguido de perto pelo **Xbox 360 (X360)**, **PlayStation 3 (PS3)** e **Nintendo Wii**.

### 4. Análise de Mercado Regional: Diferença Cultural de Consumo

```python
generos_regiao = df_clean.groupby('Genre')[['NA_Sales', 'EU_Sales', 'JP_Sales']].sum()

generos_regiao.plot(kind='bar', figsize=(14, 6))
plt.title('Preferência de Gêneros por Região (América do Norte vs Europa vs Japão)', fontsize=14, fontweight='bold')
plt.xlabel('Gênero do Jogo')
plt.ylabel('Vendas (Mi de unidades)')
plt.xticks(rotation=45)
plt.legend(['América do Norte (NA)', 'Europa (EU)', 'Japão (JP)'])
plt.show()
```

**💡 Insight 2:** Enquanto a **América do Norte** e a **Europa** consomem majoritariamente jogos de **Ação (Action)** e **Esportes (Sports)**, o **Japão (JP)** possui uma preferência marcante pelo gênero de **RPG (Role-Playing)**, superando qualquer outro segmento no país.
